# Self learn of all 4 OOPs concepts using streaming platform example

## 1. Encapsulation: 

* data and functions inside a class
* Restricting direct access to sensitive data.
* Use private variable "__" for encapsulation

## 2.Abstraction: 

* Show only essential features to the user
* Hide internal working details
* What does @abstractmethod does is:
     * It forces child classes to implement that method.
     * It ensures a common interface across all subclasses and prevents creating incomplete objects.

## 3.Inheritance (Code Reusability):

* One class can use properties and methods of another class
* Parent → Child relationship
* No need to rewrite the classes everytime where it required
* It helps for code reuseability



## 4.Polymorphism (One Interface, Many Forms)

* Polymorphism means it has same method across the mentioned classes but has different behavior
* “Poly” = many
* “Morphism” = forms 

## Example code: 

```python
from abc import ABC, abstractmethod

# ABSTRACTION:

class User(ABC):
    def __init__(self, username, password):
        self.username = username
        self.__password = password   # Encapsulation

    def login(self):
        entered = input("Enter password: ")
        if entered == self.__password:
            print("Login Successful")
            return True
        else:
            print("Wrong Password")
            return False

    @abstractmethod
    def watch_video(self):
        pass

    @abstractmethod
    def subscription_type(self):
        pass
        
# INHERITANCE:

class FreeUser(User):
    def watch_video(self):
        print("Playing video...")
        print("Showing Ads...")

    # POLYMORPHISM:
    def subscription_type(self):
        print("Free Subscription")

class PremiumUser(User):
    def watch_video(self):
        print("Playing video without Ads")

    # POLYMORPHISM:
    def subscription_type(self):
        print("Premium Subscription")

    def download_video(self):
        print("Video Downloaded Successfully")

# Platform Interface
def streaming_platform(user):
    if not user.login():
        return
    while True:
        print("1. Watch Video")
        print("2. Show Subscription Type")
        print("3. Download Video (Premium Only)")
        print("4. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            user.watch_video()

        elif choice == "2":
            user.subscription_type()

        elif choice == "3":
            if isinstance(user, PremiumUser):
                user.download_video()
            else:
                print("Upgrade to Premium to Download")

        elif choice == "4":
            print("Thank You for Using Platform")
            break

        else:
            print("Invalid Choice")

# DRIVER CODE:

print("Select User Type:")
print("1. Free User")
print("2. Premium User")

choice = input("Enter choice: ")

if choice == "1":
    user = FreeUser("Rahul", "1234")
elif choice == "2":
    user = PremiumUser("Amit", "5678")
else:
    print("Invalid Selection")
    exit()

streaming_platform(user)
```

### Inference from the code: 

* **Encapsulation** protects user password.
* **Abstraction** hides implementation details of video streaming.
* **Inheritance** allows FreeUser and PremiumUser to reuse common login features.
* **Polymorphism** allows watch_video() to behave differently depending on subscription type.
